## 1. Import Libraries
Import `sqlite3` for database connection and `pandas` for reading CSV files and displaying results.

```python
# Import libraries
import sqlite3  # SQLite database library
import pandas as pd  # Data handling and display
```

## 2. Mount Google Drive and Read CSV Files
Upload `emp.csv` and `dept.csv` into Google Colab.

```python
# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Read CSV files from Google Drive
emp_df = pd.read_csv("/content/drive/My Drive/Applied Database Technologies/emp.csv")
dept_df = pd.read_csv("/content/drive/My Drive/Applied Database Technologies/dept.csv")
```

## 3. Create SQLite Database
Create a new SQLite database and a cursor to execute SQL statements.

```python
# Connect to a new SQLite database
conn = sqlite3.connect("company.db")
cursor = conn.cursor()
```

## 4. Load CSV Data into Tables
Read the CSV files with `pandas` and load them into SQLite tables using `to_sql()`.

```python
# Load DataFrames into SQLite tables
emp_df.to_sql("emp", conn, if_exists="replace", index=False)
dept_df.to_sql("dept", conn, if_exists="replace", index=False)

print("Tables created and data loaded successfully.")
```

## 5. SQL Queries (Chapter 1.1–1.5)
Execute SQL queries from SQL Cookbook Chapter 1, Problems 1.1 through 1.5.

### Problem 1.1 – Retrieving All Rows and Columns
Use `SELECT *` to retrieve all data from the `emp` table.

```python
# Problem 1.1: Select all rows and columns
query_1 = """
SELECT *
FROM emp;
"""

pd.read_sql_query(query_1, conn)
```

### Problem 1.2 – Retrieving a Subset of Rows
Use the `WHERE` clause to retrieve only employees in department 10.

```python
# Problem 1.2: Filter rows using WHERE
query_2 = """
SELECT *
FROM emp
WHERE deptno = 10;
"""

pd.read_sql_query(query_2, conn)
```

### Problem 1.3 – Finding Rows That Satisfy Multiple Conditions
Use `WHERE` with `AND` and `OR` to filter rows based on multiple conditions.

```python
# Problem 1.3: Multiple conditions with AND/OR
query_3 = """
SELECT *
FROM emp
WHERE deptno = 10
   OR comm IS NOT NULL
   OR sal <= 2000 AND deptno = 20;
"""

pd.read_sql_query(query_3, conn)
```

### Problem 1.4 – Retrieving a Subset of Columns
Select only specific columns (`ename`, `deptno`, `sal`) instead of all columns.

```python
# Problem 1.4: Select specific columns
query_4 = """
SELECT ename, deptno, sal
FROM emp;
"""

pd.read_sql_query(query_4, conn)
```

### Problem 1.5 – Providing Meaningful Names for Columns
Use `AS` to create readable column aliases for `sal` and `comm`.

```python
# Problem 1.5: Rename columns using aliases
query_5 = """
SELECT sal AS salary,
       comm AS commission
FROM emp;
"""

pd.read_sql_query(query_5, conn)
```

## 6. Close Connection
Close the SQLite database connection to free resources.

```python
# Close the database connection
conn.close()
print("Database connection closed.")
```
